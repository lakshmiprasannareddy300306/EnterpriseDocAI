```python
import pandas as pd
import numpy as np
import zipfile
import os

zip_path = '/content/online+retail+ii.zip'
extract_path = '/content/'

# Extract the zip file
if os.path.exists(zip_path):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_path)
    print("Extraction complete.")
else:
    print(f"Error: {zip_path} not found.")

# Look for any .xlsx file extracted
files = os.listdir(extract_path)
excel_files = [f for f in files if f.endswith('.xlsx')]

if excel_files:
    filename = os.path.join(extract_path, excel_files[0])
    print(f"Loading: {filename}")
    excel = pd.ExcelFile(filename)
    print("Available sheets:")
    print(excel.sheet_names)
else:
    print("No Excel files found in the extracted content.")
```

```python
df = pd.concat(
    [
        pd.read_excel(filename, sheet_name=sheet)
        for sheet in excel.sheet_names
    ],
    ignore_index=True
)

print("Dataset loaded successfully!")
```

```python
df_clean = df.copy()

print("Working dataset created!")
print("Rows:", len(df_clean))
```

```python
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df_clean.columns.tolist())
```

```python
df_clean['customer_id']
```

```python
duplicate_count = df_clean.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
df_clean = df_clean.drop_duplicates()

print("Duplicates removed!")
print("Remaining rows:", len(df_clean))
```

```python
df_clean['invoicedate'] = pd.to_datetime(
    df_clean['invoicedate'],
    errors='coerce'
)

df_clean['customer_id'] = df_clean['customer_id'].astype('Int64')

df_clean['quantity'] = pd.to_numeric(
    df_clean['quantity'],
    errors='coerce'
)

df_clean['price'] = pd.to_numeric(
    df_clean['price'],
    errors='coerce'
)

print(df_clean.dtypes)
```

```python
df_clean['invoice'] = df_clean['invoice'].astype(str).str.strip()

df_clean['transaction_type'] = np.select(
    [
        df_clean['invoice'].str.startswith('C'),
        df_clean['quantity'] < 0
    ],
    [
        'Cancellation',
        'Return_or_Adjustment'
    ],
    default='Regular'
)

print(df_clean['transaction_type'].value_counts())
```

```python
df_clean['transaction_value'] = (
    df_clean['quantity'] * df_clean['price']
)

df_clean[['invoice', 'quantity', 'price',
          'transaction_value', 'transaction_type']].head(10)
```
